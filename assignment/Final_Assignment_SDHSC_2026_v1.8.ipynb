```python
# the usual imports.
# You have to do this yourself: e.g. you are using graphing libraries etc. bring your imports here.
import pprint as pp
import json
```

```python
# this function just loads the data from files. We wrote it for you, and there is no need to understand how it does it.
def load_json_file_named(file_name):
    try: 
        loaded_data = []
        file_location = f"{file_name}"
        with open(file_location, 'r') as file: # or f"data/{file_name}" depending on your files
            loaded_data =  json.load(file)
    except OSError as e:
        print(f"Error. Does the file exist in this folder? {file_location}\n\n {e}")
    return loaded_data
```

Now let's load the data from files:


```python
boards_info = load_json_file_named('../data/nhs_boards_26.json')
covid_days = load_json_file_named('../data/nhs_covid_days_26.json')

# check if data loaded:
print(f"Boards: {len(boards_info)}")
print(f"Days: {len(covid_days)}")
```
